```python
# --- Convergence Checks ---

# 1. Visual Inspection with Trace Plots
# This generates plots for each parameter showing the sampled values across iterations
# for each chain (left) and the posterior density estimate (right).
# What to look for:
#   - Left plot (Trace): Chains should look like "fuzzy caterpillars" mixing well,
#     centred around a stable value, without significant drifts or trends.
#     Multiple chains should overlap each other.
#   - Right plot (Density): Posterior density estimates from different chains
#     should largely agree.
print("Generating Trace Plots...")
try:
    trace_plot = az.plot_trace(data)
    # Depending on your environment (script vs notebook), you might need plt.show()
    plt.show()
    print("Trace plots generated. Please inspect visually.")
except Exception as e:
    print(f"Could not generate trace plots: {e}")
    print("Check if the InferenceData object 'data' is structured correctly and contains posterior samples.")

print("\n" + "="*50 + "\n")

# 2. Quantitative Checks with R-hat (r_hat) and ESS
# This calculates summary statistics including r_hat and effective sample size (ESS).
# What to look for:
#   - 'r_hat': Should be close to 1.0 (e.g., < 1.01 or < 1.05). Values much
#      larger than 1 indicate chains haven't converged to the same distribution.
#   - 'ess_bulk' and 'ess_tail': Effective Sample Size. Higher is better. Low ESS
#      (e.g., < 100 or < few hundred, depending on needs) suggests high
#      autocorrelation and inefficient sampling, meaning you might need more
#      samples or have model issues. ess_bulk relates to the centre of the
#      distribution, ess_tail relates to the tails (important for credible intervals).
print("Calculating Summary Statistics (including R-hat and ESS)...")
try:
    summary_df = az.summary(data)
    print(summary_df)

    # You can programmatically check r_hat values
    max_rhat = summary_df['r_hat'].max()
    print(f"\nMaximum R-hat value: {max_rhat:.4f}")


    rhat_threshold = 1.05 # Define your threshold
    # Filter the DataFrame for rows where r_hat exceeds the threshold
    high_rhat_params = summary_df[summary_df['r_hat'] > rhat_threshold]

    if not high_rhat_params.empty:
        print(f"\nWARNING: Found {len(high_rhat_params)} parameters with r_hat > {rhat_threshold}.")
        print("Potential convergence issues detected. Details below:")
        print("----------------------------------------------------")

        # Iterate through the filtered DataFrame (row by row)
        for parameter_name, stats_row in high_rhat_params.iterrows():
            print(f"Parameter: {parameter_name}")
            # Print the Series corresponding to this parameter's stats
            # This automatically formats it nicely column-wise
            print(stats_row)
            print("-" * 20) # Add a separator line

        print("----------------------------------------------------")
    else:
        # This message is printed if no parameters exceed the threshold
        print(f"\nAll R-hat values seem acceptable (<= {rhat_threshold}).")

    # You can also check minimum ESS values
    min_bulk_ess = summary_df['ess_bulk'].min()
    min_tail_ess = summary_df['ess_tail'].min()
    print(f"\nMinimum Bulk ESS: {min_bulk_ess:.1f}")
    print(f"Minimum Tail ESS: {min_tail_ess:.1f}")
    ess_threshold = 100 # Example threshold
    if min_bulk_ess < ess_threshold or min_tail_ess < ess_threshold:
         print(f"WARNING: Minimum ESS is below {ess_threshold}, indicating potential issues with sampling efficiency or autocorrelation.")
         print("Parameters with low ESS:")
         print(summary_df[(summary_df['ess_bulk'] < ess_threshold) | (summary_df['ess_tail'] < ess_threshold)])
    else:
        print(f"Minimum ESS values seem acceptable (>= {ess_threshold}).")

except Exception as e:
    print(f"Could not calculate summary statistics: {e}")
    print("Check if the InferenceData object 'data' is structured correctly.")

print("\n" + "="*50 + "\n")
print("Convergence checks complete. Review the plots and summary table.")
```

### Preprocessing the data

We have a bunch of separate timeseries for each pulsar.

We want to bring these into a single data object that the Kalman filter can iterate through.

```python
from gravitational_waves import hellings_downs

# Get the separation angles and compute HD correlation
ra = pulsar_metadata["RA"].to_numpy(dtype=float)
dec = pulsar_metadata["DEC"].to_numpy(dtype=float)
angular_separation_matrix = LoadWidebandPulsarData.pairwise_angular_separation(ra, dec)
hd_correlation_matrix = hellings_downs(angular_separation_matrix)

# Post-process the residuals
processed_pulsar_residuals = LoadWidebandPulsarData.post_process_residuals(pulsar_residuals)

print("Total length of the data is ", len(processed_pulsar_residuals))
print("Total number of pulsars is ", len(pulsar_metadata))
```
